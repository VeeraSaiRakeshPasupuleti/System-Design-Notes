# System Design

Imagine you are building a simple application that works perfectly for 100 users. Now the application suddenly has to serve **1 million users** at the same time. Requests start increasing, the database becomes overloaded, response times increase, and eventually the system **may stop working**.

This is where **System Design** becomes important.

Learning System Design can feel overwhelming at first. There are databases,APIs, caching, load balancing, message queues, scalability, consistency, distributed systems and many other concepts to understand. The challenging is not only learning these components individually, but also understanding **WHEN, WHY and HOW** to use them together.

This 20-day roadmap is designed to take you from **the fundamentals of System Design** to **designing real-world scalable systems**, step-by-step.


Instead of trying to learning everything at once, we will build our knowledge progressively:


**Fundamentals --> Core Components --> Scalability --> Distributed Systems --> Architecture Patters --> Real-World System Design**

### What you will Learn

Over the next 20 days, we will cover:

* **System Design Fundamentals** - requirements, contraints, scalability, availability, reliability and performace
* **Networking and APIs** - HTTP, HTTPS, DNS, TCP, RESPT APIs, and communication between services.
* **Databases** - SQL vs NoSQL, indexing, replication, partitioning and database scaling.
* **Core Infrastructure Components** - load balancers, caching, CDNs, reverse proxies and meessage queues.
* **Scalability and Reliability** - horizontal scaling, vertical scaling, fault tolerance, redundancy and high availability.
* **Distributed Systems** - consistency, CAP theorem, distributed transactions and communicational patterns.
* **Architecture Patterns** - monoliths, microservices, event-driven architecture and asynchronous processing.
* **Real-World System Design** - applying everything we have learned to systems such as URL shortner, notification stsrems, chat applications, other large scale applications.

### How this Roadmap Works

Each day focuses on a specific concept and follows a consistent learning approach:

**Concept --> Why it matters --> How it works --> Architecture diagram --> Real-world example --> Design considerations**

The goal is not to memoize the system-design diagrams, Instead we will learn **how to think about a system**, identify its requirements, make architectural decisions, understand trade-offs, and gradually arrive at a scalable design.

By the end of Day 20, you should have a structured understanding of the major building blocks used in High-Level Design (HLD) and be able to approach a system design problem systematically.

### HLD vs LLD
This is one of the first concept that you should understand.

**HLD** - High-Level Design describes the overall architechture of the system.

It focuses on Services, Databases, APIs, Load Balancers, Caches, Message Queues, External Systems, Scalanility, Reliability etc.,

The HLD defines <br/> 
**What component do we need and how do they communicate?**

**LLD** - Low-Level Design goes deeper into the implementation of individual components.<br /> 
For example, suppose **HLD contains Order Service**<br /> 
LLD may describe Order class, OrderService, OrderRepository, OrderDatabase.<br /> 

LLD focuses on Classes, Interfaces, Methods, Objects, Design patterns, Database Schemas, Detailed algorithms.<br /> 

A useful way to remember the difference between HLD and LLD:<br /> 
HLD<br /> 
**"What components do we need?"**<br /> 

LLD<br /> 
**"How do we implement those components?"**<br /> 


### Functional Requirements
Before designing anything, we need to understand **what the system should do.**
These are called **Functional Requirements.**

For an e-commerce system Functional Requiremnts could be as follow:
* User can register.
* User can login.
* User can search products.
* User can view products.
* User can add products to cart.
* User can place an order.

These describe **system behaviour.**

### Non-Functional Requirements
Functional requirements tell us:
**What should the system do?**

Non-functional requirements tells us:
**How well should system do it?**

**Examples:** Performance, Scalability, Availability, Reliability, Security, Maintainability, Durability

Functional: User can upload a file
Non-functional: 
1. Upload should complete within 3 seconds.
2. System should support 100000 concurrent users.
3. Uploaded files should not be lost.
4. System should remain avaiable during server failures.

### Important Non-Functional Requirements

**Scalability** - Scalability means ability of a system to handle increasing workload by adding resources or changit its architecture.
A scalable system should be able to grow without requiring a complete rewrite or a manual intervention.

**Availability** - describes how often a system is operational and accessible. For example, 99% availability, 99.9% availability etc.,

**Reliability** - is about whether a system continues to behave correctly and consistently including when failures occur. 

**Performance** - describes how efficiently a system responds to requests. Two common metrics that calculate performance are **Latency** and **Throughput**

**Latency** -  How long does once request took to respond.
**Throughput** - How many requests can the system process in a period of time.

### 2. Functional Requirements
Functional requirements describe **the actual operations the systems** must support

For an e-commerce application, these migh include:

=> Register
=> Login
=> Search Product
=> Add to cart
=> Place Order
=> Make Payment
=> Track Order

Functional requirements determine much of the system's behaviour and eventually influence: APIs, Services, Data models, Database operations, Communication flows etc.,

If the required functionality is unclear, the architecture will also be unclear.

### 3. Non-Functional Requirements
Non-functional requirements describe the q**ualities and contraints of the system** rather than its individual features.

**Typical examples include:** Scalability, Availability, Performance, Reliability, Security, Durability, maintainability etc.,

For Example, for a payment system the requirements might be:

Availability --> 99.99%
Latency --> 95% of requests < 200ms
Reliability --> No duplicate payments
Security --> Data encrypted in transit and at rest
Scalability --> Support increasing transaction volume

### 4. Capacity Estimation
Capacity estimation is the process of approximating the amount of **traffic, storage, bandwidth and computing capacity** a system will require.

Instead of saying: **The application will have lots of traffic.**

We try to reach statement such as: **The aplpication needs to handle approximately 5000 requests per second.**

These estimates influence architectural descisions such as:
* Number of application servers.
* Database capacity.
* Storage requirements.
* Network requirements.
* Whether caching is necessary.
* Whether database partitioning is necessary.

Without an estimation, architecture decisions are largely guesses.

### 5. Requests Per Second - RPS
RPS represents the number of requests a system receives or processes every second.

It is one of the most useful numbers when reasoning about system capacity.

Suppose we know **100 million requests/day**, that number alone doesn't immediately tell you how much capacity the application needs. Converting it to RPS makes the workload easier to reason about.


### 6. Peak Traffic
Peak traffic is the highest or near-highest traffic level that the system is expected to handle

Designing only for the average traffic can leave the system vulnerable to traffic spikes.

For example:
Average = 1000 RPS
Peak = 5000 RPS

If the architecture can process only 1000 RPS, it may degrade severly during the peak period. 


A simple estimation approach is:

Peak RPS = Average RPS X Peak factor

For example, assuming a 3x peak factor

Average RPS=1000
Peak factor =3

Peak RPS= 3000

Note: The **3× factor is only an illustrative assumption**, not a universal rule. In real systems, historical traffic patterns or business expectations should determine the appropriate peak factor.

### 7. Read-to-Write Ratio
The read/write ratio describes how often data is read compared with how often ot is written.

For example, **90% Reads and 10% Writes**.
represents a read-heavy workload.

Different workload patterns can lead to different architectural choices.

For read-heavy workloads, we have techniques such as:
* Caching
* Read Replicas

For Write-heavy workloads, the architecture may need to emphasize:
* Efficient Writes
* Partitioning
* Batching
* Queues
* Write-Optimized Storage

Suppose the system receives **10,000 RPS** and 
Reads = 9000
Writes = 1000

Then, Read:Write = 9:1

### 8.Storage Estimation
Storage estimation determines how much data the system will generate and retain.

Storage requirements influence
* Database selection
* Storage architecture
* Backup strategy
* Replication
* Cost
* Data retention

### 9. Bandwidth Estimation
Bandwidth estimation determines how much data needs to move through the network.

This becomes especially important for system handling: 
**Videos, Images, Files, Streaming, Large API responses.**

A system can have enough CPU and database capacity and still become network-bound.

For example, serving large videos generates vastly more network traffic than returning a small JSON response.

Suppose, RPS = 1000
Average response = 200KB

then 1000 x 200 KB = 200000 KB/sec = **200 MB/sec**
That's subtantial network traffic.

### 10. Back-of-the-envelope Estimation
Back-of-the-envelope estimation is a quick way to approximate the scale of a system using simple arithmetic and explicit assumptions.

The goal isn't perfect accuracy.

The goal is to determine whether the system is roughly dealing with:
10 RPS? Or
1000 RPS? Or
1 million RPS?

Suppose you're asked.
**How many application servers would you need?**

Instead of guessing: 20 servers

reason through:
Expected RPS -> Capacity of one server -> Required server count

### 12. From Requiremnts to Architecture
Capacity estimation isn't a seperate mathematical exercise. Its purpose is to drive architectural decisions.

Suppose your calculations reveal:
Very high read traffic ---> We can go with caching
large file storage requirements ---> We can go with object storage
Traffic is too large ---> Horizontal Scaling
